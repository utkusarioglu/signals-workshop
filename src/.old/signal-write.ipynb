```python
import numpy as np
import os
import scipy
import os
from pathlib import Path


WAV_ABSPATH = Path(os.environ["ARTIFACTS_ABSPATH"]) / "equal-perfect.wav"
DUR = 2
SR = 44100
BASE = 400
WAVE_TYPE = "square"
FREQS = {
    "equal": [BASE, BASE * 2 ** (4 / 12), BASE * 2 ** (7 / 12)],
    "perfect": [BASE, BASE * (5 / 4), BASE * (3 / 2)],
}


def create_sig(freqs: list[int], duration: int, wave_type: str):
    time = np.linspace(
        0,
        duration,
        int(SR) * duration,
        endpoint=False,
    )
    sig = []
    match wave_type:
        case "sine":
            method = np.sin
        case "sawtooth":
            method = getattr(scipy.signal, "sawtooth")
        case "square":
            method = getattr(scipy.signal, "square")
    for freq in freqs:
        sig.append(method(2 * np.pi * freq * time))
    sig = np.sum(sig, axis=0)
    sig_max = np.max(np.abs(sig))
    sig = sig / sig_max
    return sig


sigs = []
for i in range(4):
    for c in FREQS.values():
        sigs.append(create_sig(c, DUR, WAVE_TYPE))

sig = np.concatenate(sigs)

pcm = np.int16(sig * (2**15 - 1))
scipy.io.wavfile.write(WAV_ABSPATH, SR, pcm)
```
