```python
from pythonosc.udp_client import SimpleUDPClient
from time import sleep
import math
from music21 import note, scale, stream, scale
```

```python
bm = scale.MinorScale("B-")
freqs = []
client = SimpleUDPClient("192.168.1.151", 57121)
for d in list(range(1, 8)) + list(range(6, 1, -1)):
    freqs.append(bm.pitchFromDegree(d).frequency)
freqs = [f - min(freqs) for f in freqs]
freqs = [f / max(freqs) for f in freqs]
# freqs = [f + 0.3 for f in freqs]
# freqs = [math.pow(f, 3) for f in freqs]
for i in range(100):
    for f in freqs:
        print(f)
        client.send_message("/phase", [1, f])
        sleep(60 / 60)
```
