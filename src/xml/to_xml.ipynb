```python
import os
import dicttoxml
import xmltodict
import yaml
import io
from pathlib import Path
from music21 import converter

FILENAME = "evocation.musicxml.yaml"
TEMP_FILENAME = "temp.musicxml"
```

```python
asset_abspath = (
    Path(os.environ.get("ASSETS_ABSPATH", "")) / "MusicXmlYaml" / FILENAME
)


test = {
    "yaml": open(asset_abspath, "r").read(),
}
test["dict"] = yaml.safe_load(test["yaml"])
test["xml"] = xmltodict.unparse(test["dict"])

with open(TEMP_FILENAME, "w") as f:
    f.write(test["xml"])
# test["xml_stream"] = io.StringIO(test["xml"])
score = converter.parse(TEMP_FILENAME, format="xml")
score.show()
```
