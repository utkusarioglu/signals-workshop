# Sonification

This notebook creates sonification for some typical algorithms.

```python
from os.path import join
from utils.supercollider import SuperCollider
from utils.chain import Chain

HOST = "supercollider"
supercollider = SuperCollider(
    join("../..", "artifacts", "scd_runtime.json"), HOST
)
```

```python
def sonify(chain, cb, base, duration, amp, unsorted):
    for note in unsorted:
        chain.note(duration, base + note, min(amp + 0.1, 1))
    chain.rest(0.5)
    response = cb(unsorted, lambda f: chain.note(duration, base + f, amp))
    chain.rest(0.5)
    for note in response:
        chain.note(duration, base + note, min(amp + 0.1, 1))
    chain.stream()
    return response
```

```python
from sonification.merge_sort import recursion

unsorted = [11, 0, 2, 9, 7, 4, 5, 12]

chain = Chain(supercollider, "piano", True)
chain.clear()
response = sonify(chain, recursion, 60, 0.3, 0.5, unsorted)
print("response:", response)
```
