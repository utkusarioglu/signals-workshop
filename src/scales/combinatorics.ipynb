```python
import itertools
from scale import Scale, Tone, Family
```

```python
all_states = itertools.product([0, 1], repeat=11)
for a in enumerate(all_states):
    print(a)
```

```python
steps = 11
notes = 7


def combinations(steps: int, notes: int):
    """Produce all possible scales
    Creates all possible scales with the given step and note counts
    steps -- number of steps per octave. Ex: 11 for 12-TET.
    notes -- number of notes in scale. Ex: 7 for major, minor scales.
    """
    return itertools.combinations(range(1, steps + 1), notes - 1)


for combination in enumerate(combinations(11, 7)):
    print(combination)
```

```python
# degrees_list = list(combinations(11, 7))
# major_mode_index = 336
# major_mode_degrees = degrees_list[major_mode_index]
# major_scale = Scale(major_mode_index, major_mode_degrees)


def add_scale(root: Tone, scale: Scale):
    current = root
    for i in range(len(scale)):
        new_tone = Tone(root, i, scale.intervals[i], scale.degrees[i])
        new_tone = current.addNext(new_tone)
        current = new_tone
    current.setIndex(scale.index)


root = Tone(None, 0, 0, 0)
tone_count = 11
note_count = 7
for index, degrees in enumerate(combinations(tone_count, note_count)):
    scale = Scale(index, degrees, tone_count)
    add_scale(root, scale)
```

```python
def find_index(root: Tone, degrees: tuple) -> int:
    curr = root
    for degree in degrees:
        n = curr.getByDegree(degree)
        if n is not None:
            curr = n
    return curr.index
```

```python
aeolian_deg = (2, 4, 5, 7, 9, 11)
aeolian_index = find_index(root, aeolian_deg)
aeolian_sc = Scale(aeolian_index, aeolian_deg, tone_count)
```

```python
f = Family(aeolian_sc)
family_intervals = f.getIntervals()
family_intervals
```
